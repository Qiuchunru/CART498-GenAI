```python
# CART498 - Assignment 1: Surrealistic Collage with Diffusers
# Run this in a Google Colab notebook with a GPU runtime
# (Runtime > Change runtime type > GPU)

# ---- 1. Install dependencies (run once in Colab) ----
# !pip install diffusers transformers accelerate torch --quiet

import torch
from diffusers import StableDiffusionPipeline
import os

# ---- 2. Load the model ----
MODEL_ID = "runwayml/stable-diffusion-v1-5"

pipe = StableDiffusionPipeline.from_pretrained(
    MODEL_ID,
    torch_dtype=torch.float16 if torch.cuda.is_available() else torch.float32,
)

pipe = pipe.to("cuda" if torch.cuda.is_available() else "cpu")

# ---- 3. Prompts ----
# Three versions: Daytime / Dusk / Night
# The same seed is used to keep the visual style and composition consistent.

prompts = {
    "1": (
        "An elegant young woman in a vintage dark dress and long gloves, "
        "walking alone along a quiet cobblestone European street in soft daylight, "
        "beside an enormous ancient tree growing between old Parisian buildings, "
        "a small wooden door mysteriously embedded in the trunk of the tree, "
        "the door slightly open with warm light glowing from inside, "
        "ordinary people and architecture contrasting with the impossible door, "
        "subtle dreamlike atmosphere, Magritte-inspired surrealism, "
        "elegant fashion illustration, highly detailed, oil painting style"
    ),

    # --------------------------------------------------------
    # Prompt 2 — Dusk: The house grows from the tree
    # --------------------------------------------------------
    "2": (
        "The same elegant young woman in the same vintage dark dress and long gloves, "
        "standing on the same cobblestone European street at dusk, "
        "beside the same enormous ancient tree, "
        "the small wooden door in the trunk now surrounded by wooden windows, "
        "tiny balconies, staircases and fragments of rooms growing naturally "
        "from the branches, as if an entire house is slowly emerging from the tree, "
        "warm glowing windows, soft evening light, glowing street lamps, "
        "dreamlike and mysterious atmosphere, Magritte-inspired surrealism, "
        "elegant fashion illustration, highly detailed, oil painting style"
    ),

    # --------------------------------------------------------
    # Prompt 3 — Night: The tree becomes a floating house
    # --------------------------------------------------------
    "3": (
        "The same elegant young woman in the same vintage dark dress and long gloves, "
        "standing on the same cobblestone European street at night, "
        "looking up at the enormous ancient tree that has transformed into "
        "a gigantic surreal house growing from its branches, "
        "entire rooms, staircases, balconies and glowing windows floating "
        "among the branches, parts of the house extending impossibly into the night sky, "
        "the roots lifting slightly above the ground as if gravity has disappeared, "
        "moonlit sky, glowing street lamps, mysterious dreamlike atmosphere, "
        "Magritte-inspired surrealism, elegant fashion illustration, "
        "highly detailed, oil painting style"
    ),
}


# ---- 4. Generate and save images ----
os.makedirs("A1_output", exist_ok=True)

# Fixed seed = visual consistency across the three images
generator = torch.Generator(device=pipe.device).manual_seed(498)

for idx, prompt in prompts.items():

    image = pipe(
        prompt,
        num_inference_steps=30,
        guidance_scale=7.5,
        generator=generator,
    ).images[0]

    out_path = f"A1_output/prompt_{idx}.jpg"
    image.save(out_path)

    print(f"Saved {out_path}")

print("Done.")
print("Images are saved in the A1_output folder.")
```
